## Data Types

## Missing Values

Missing values are reviewed to identify variables that may require cleaning, imputation, or exclusion in later stages of the project.

```python
batting_missing = batting.isnull().sum().sort_values(ascending=False)
salary_missing = salary.isnull().sum().sort_values(ascending=False)

print("Missing values in batting dataset:")
display(batting_missing[batting_missing > 0])

print("Missing values in salary dataset:")
display(salary_missing[salary_missing > 0])
```

## Duplicate Records

Duplicate rows are identified to ensure data integrity before performing any preprocessing or data integration tasks.

```python
print("Duplicate rows in batting dataset:", batting.duplicated().sum())

print("Duplicate rows in salary dataset:", salary.duplicated().sum())
```
